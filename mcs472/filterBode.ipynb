# The Bode plot of a filter

```julia
using Plots
```

```julia
RC = 100/3
```

```julia
LC = 1/9
```

```julia
H(s) = ((1/RC)*s)/(s^2 + s/RC + 1/LC)
```

```julia
maxomega = 6
omegas = 0.01:0.01:maxomega
```

```julia
I = Complex(0,1)
```

```julia
gains = [abs(H(I*w)) for w in omegas];
phases = [angle(H(I*w)) for w in omegas];
```

```julia
gainplot = plot(omegas, gains, xlabel="omega", ylabel = "gain",
                label = "", legend=(0.7,0.5))
phaseplot = plot(omegas, phases, xlabel="omega", ylabel = "phase",
                 label = "", legend=(0.7,0.5))
plot(gainplot, phaseplot)
```
