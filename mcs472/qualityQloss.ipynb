# a quality loss function

The quality loss function derivated in the second lecture of MCS 472 is piecewise quadratic.

If a shirt is too small, for $\Delta m^- = 0.5$cm, the loss is $L^- = \$40$.

If a shirt is too wide, for $\Delta m^+ = 1$cm, the loss is $L^+ = \$20$.

Let us take $m = 40$cm as the neck size of the buyer.

The size of the produced shirt is denoted by $y$.

The piecewise quadratic quality loss function $L(y)$ depends on two parameters:

$$
    L(y) = 
    \left\{
       \begin{array}{ll}
          k^+ (y-m)^2 & \mbox{for } y \geq m, \\
          k^- (y-m)^2 & \mbox{for } y < m.
       \end{array}
    \right.
$$

We have set $m = 40$ and via interpolation, we derived in the second lecture: $k^+ = 20$ and $k^- = 160$.

The numerical quality loss function $L(y)$ consists of four pieces:

$$
   L(y) = 
   \left\{
      \begin{array}{ll}
         20 & \mbox{for } y \geq 41 \\
         20(y - 40)^2 & \mbox{for } 40 \leq y < 41 \\
         160(y - 40)^2 & \mbox{for } 39.5 < y < 40 \\
         40 & \mbox{for } y \leq 39.5
      \end{array}
   \right.
$$

To make a plot, we define a Julia function for $L$.

```julia
"""
    Defines the L(y) function above.
"""
function loss(y)
    if y >= 41
        return 20
    elseif y >= 40
        return 20*(y-40)^2
    elseif y > 39.5
        return 160*(y-40)^2
    else
        return 40
    end
end
```

To make a plot, we define a range for $y$, we let $y$ range from 38 to 42.
The function ``loss`` is evaluated at all points in this range.

```julia
r = 38:0.01:42
v = [loss(t) for t in r]
plot(r,v,label="L(y)")
```
