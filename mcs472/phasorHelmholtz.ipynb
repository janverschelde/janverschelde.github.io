# Periodic Steady State

*How deep do we have to dig to store something a constant temperature?*

We look at the solution of the heat equation where the boundary condition is $\sin(\omega t)$, for all values of $x$ at the surface.

```julia
using SymPy
```

```julia
using Plots
```

## 1. A Symbolic Computation

With ``SymPy`` we can obtain a solution of the decoupled problem.

```julia
x = Sym("x")
```

```julia
omega = Sym("omega")
```

```julia
y = SymFunction("y")
```

```julia
ode = diff(diff(y(x), x),x) - sympy.I*omega*y(x)
```

```julia
dsolve(ode, y(x))
```

## 2. Running a Simulation

The main independent constants are below.

```julia
alpha = 1.0E-6 # thermal diffusion constant
```

```julia
P = 24*3600 # 24-hour oscillation period measured in seconds
```

```julia
T0 = 10 # temperature at the surface in Celsius
A = 10 # maximum amplitude variation
```

```julia
omega = 2*pi/P
```

```julia
delta = sqrt(2*alpha/omega) # depth of penetration
```

```julia
T(x, t) = T0 + A*exp(-x/delta)*cos(omega*t - x/delta)
```

```julia
T1(x) = T(x, 1*3600)
plot(0:0.01:delta*5, T1, xlabel="depth", ylabel="temperature", label="the 1st hour", ylims=[0, 20])
```
