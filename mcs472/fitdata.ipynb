# An Example of Fitting Data

With this example we add random noise to a line.

```julia
using Plots
```

```julia
x = -0.1:0.1:1.1
y = 0.5*x .+ 1
plot(x,y,label="y=0.5*x + 1",legend=:topleft)
```

```julia
ynoisy = [y[i] + 0.2*randn() for i=1:length(y)]
scatter!(x[2:length(x)-1],ynoisy[2:length(x)-1],marker=(:red),label="data",legend=:topleft)
```

```julia
A = ones(length(x),2)
A[:,2] = x
A
```

```julia
c = A\ynoisy
```

```julia
lbl = string("y=", string(c[2]), "*x+", string(c[1]))
fity = c[1] .+ c[2]*x
plot!(x,fity,label=lbl)
```
