## 1. Formulas for the Laplace Transform

Our text book contains tables with the Laplace transforms for most of the common functions.  SymPy offers the functions ``sympy.laplace_transform`` and ``sympy.inverse_laplace_transform`` to compute the Laplace transform and its inverse.

```julia
@syms s, t, a::positive, n::integer
```

```julia
lap1 = sympy.laplace_transform(1, t, s)
```

```julia
invlap1 = sympy.inverse_laplace_transform(lap1[1], s, t)
```

The $\theta(t)$ function is the Heaviside function.

```julia
laptn = sympy.laplace_transform(t^n, t, s)
```

```julia
invlaptn = sympy.inverse_laplace_transform(laptn[1], s, t)
```

```julia
lapexp = sympy.laplace_transform(exp(-a*t), t, s)
```

```julia
invlapexp = sympy.laplace_transform(lapexp[1], s, t)
```

```julia
lapcos = sympy.laplace_transform(cos(a*t), t, s)
```

```julia
invlapcos = sympy.inverse_laplace_transform(lapcos[1], s, t)
```

```julia
lapsin = sympy.laplace_transform(sin(a*t), t, s)
```

```julia
invlapsin = sympy.inverse_laplace_transform(lapsin[1], s, t)
```

```julia
lapcosh = sympy.laplace_transform(cosh(a*t), t, s)
```

```julia
invlapcosh = sympy.inverse_laplace_transform(lapcosh[1], s, t)
```

```julia
lapsinh = sympy.laplace_transform(sinh(a*t), t, s)
```

```julia
invlapsinh = sympy.inverse_laplace_transform(lapsinh[1], s, t)
```

The Laplace transform of the Dirac delta function is 1.

```julia
diracdelta = sympy.inverse_laplace_transform(1, s, t)
```

```julia
sympy.laplace_transform(diracdelta, t, s)
```

## 2. Solving a System of Differential equations

The Laplace transform of

$$
    \left\{
       \begin{array}{rcr}
          x' & = & -2 x - 9 y \\
          y' & = &    x - 2 y
       \end{array}
    \right. \quad x(0) = -1, y(0) = 1
$$

is

$$
    \left\{
       \begin{array}{rcr}
          s X + 1 & = & -2 X - 9 Y \\
          s Y - 1 & = &    X - 2 Y
       \end{array}
    \right.
$$

or

$$
    \left\{
       \begin{array}{rcr}
          (s+2) X + 9 Y & = & -1 \\
          -X + (s+2) Y & = & 1
       \end{array}
    \right.
$$

where $X(s)$ and $Y(s)$ are the Laplace transforms
of $x(t)$ and $y(t)$ respectively.

```julia
@syms X, Y
```

```julia
eq1 = (s+2)*X + 9*Y + 1
eq2 = -X + (s+2)*Y - 1
eqs = [eq1, eq2]
```

```julia
sol = sympy.solve(eqs, [X, Y])
```

Now we transform the solution into the type domain.

```julia
xt = sympy.inverse_laplace_transform(sol[X], s, t)
```

```julia
yt = sympy.inverse_laplace_transform(sol[Y], s, t)
```
