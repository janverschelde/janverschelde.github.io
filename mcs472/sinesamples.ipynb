# Sampling a sine function

This notebook illustrates the sampling of a sine function, running at one Hertz, taking samples at eight Hertz.  The unit Hertz (abbreviated by Hz) measures the number of cycles per second, so $n$ Hertz stands for $n$ cycles per second.

```julia
using Plots
```

```julia
f(t) = sin(2*pi*t)
```

In the plot below, the grid lines are set to introduce the sampling at 8 Hz.

```julia
rt = 0:0.01:1
rf = [f(x) for x in rt]
xr = 0:0.125:1
xvalues = [x for x in xr]
plot(rt, rf, label="sin(2*pi*t)", xticks=xvalues, xlabel="t in seconds")
```

```julia
# savefig("sine1Hz.png")
```

Now we sample the sine function at 8 Hertz, taking 8 samples per second.

```julia
sf = [f(x) for x in xr]
scatter!(xr, sf, label="")
```

Let us emphasize the sampling, adding vertical lines for every sample.

```julia
plot!([0.125, 0.125], [0, f(0.125)], color=:black, label="")
plot!([0.25, 0.25], [0, f(0.25)], color=:black, label="")
plot!([0.375, 0.375], [0, f(0.375)], color=:black, label="")
plot!([0.625, 0.625], [0, f(0.625)], color=:black, label="")
plot!([0.75, 0.75], [0, f(0.75)], color=:black, label="")
plot!([0.875, 0.875], [0, f(0.875)], color=:black, label="")
```
