# Separation of Variables

In the corresponding lecture on the method of the separation of variables, we considered

$$
    \frac{\partial u}{\partial t} = \frac{\partial^2 u}{\partial x^2}, \quad x \in [0,1],
$$

with initial condition

$$
    u(x, 0) = 1
$$  

and boundary conditions

$$
    u(0, t) = 0 \quad \mbox{and} \quad u(1, t) = 0.
$$

The method of separation of variables produced the solution

$$
   U(x, t) = \sum_{k=0}^\infty \frac{4}{(2k + 1)\pi} \left( e^{-(2k+1)\pi^2 t} \sin((2k+1)\pi x) \right).
$$

The purpose of this notebook is to explore the convergence of this solution.

## 1. The Fourier series approximation for one.

At $t=0$, we compute the Fourier approximation of one, for $x \in [0,1]$.

```julia
U10(x) = (4/pi)*sum([sin((2*k+1)*pi*x)/(2*k+1) for k=0:10])
```

```julia
plot(0:0.0001:1, U10, xlabel="x", label="10 terms", ylims=[-0.03,1.3], legend=(0.5,0.5))
```

## 2. plotting for t

```julia
U160t1(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.0001))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
U160t2(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.001))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
U160t3(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.01))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
U160t4(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.05))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
U160t5(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.1))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
U160t6(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.2))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
U160t7(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.3))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
U160t8(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.4))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
U160t9(x) = (4/pi)*sum([exp(-(2*k+1)^2*pi^2*(0.5))*sin((2*k+1)*pi*x)/(2*k+1) for k=0:160])
```

```julia
plot(0:0.0001:1, U160t1, xlabel="x", label="t = 0.0001", ylims=[-0.03,1.1], legend=:outertopright)
plot!(0:0.0001:1, U160t2, xlabel="x", label="t = 0.001")
plot!(0:0.0001:1, U160t3, xlabel="x", label="t = 0.01")
plot!(0:0.0001:1, U160t4, xlabel="x", label="t = 0.05")
plot!(0:0.0001:1, U160t5, xlabel="x", label="t = 0.1")
plot!(0:0.0001:1, U160t6, xlabel="x", label="t = 0.2")
plot!(0:0.0001:1, U160t7, xlabel="x", label="t = 0.3")
plot!(0:0.0001:1, U160t8, xlabel="x", label="t = 0.4")
plot!(0:0.0001:1, U160t9, xlabel="x", label="t = 0.5")
```
