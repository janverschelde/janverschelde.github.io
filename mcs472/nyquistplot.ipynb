# Nyquist plot

```julia
using Plots
```

```julia
P(s) = 1/(s-1)
```

```julia
H(s) = 1/(s-2)
```

```julia
G(s) = H(s)*P(s)
```

```julia
I = Complex(0,1)
r(omega) = G(omega*I)
omegarange = -100:0.01:100
rt = [(real(r(t)), imag(r(t))) for t in omegarange]
plot(rt, xlabel="real part", ylabel = "imag part", label = "G(w*I)")
```
