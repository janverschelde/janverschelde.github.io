A train leaves the depot empty.

At each station, on average,

* 40 people mount,

* while 20% of the riders dismount.

*Will there ever be a train long enough to accommodate the 40 people mounting at each station?*

```julia
using Plots
```

# SymPy

```julia
using SymPy
```

## 1. Solving Recursions with SymPy

```julia
n = Sym("n")
```

```julia
x = SymFunction("x")
```

```julia
eq = Eq(x(n), (4/5)*x(n-1) + 40)
```

```julia
init = Dict(x(0) => 0)
```

```julia
sol = sympy.rsolve(eq, x(n), init)
```

## 2. plotting the solution

```julia
funsol(valn) = sol.subs(n, valn)
```

```julia
funsol(30)
```

```julia
vals = [funsol(t) for t=0:20]
```

```julia
plot(0:20, vals, label="200 - 200*(0.8)^n", legend=:bottomright)
```

From the plot we can see that a train with 200 seats will suffice to accommodate the 40 people getting on the train at each stop.
