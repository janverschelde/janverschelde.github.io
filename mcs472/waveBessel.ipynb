## 1. Symbolic Manipulations

We first verify that the Bessel functions satisfy the ordinary differential equation

$$
   z^2 \frac{\partial^2 R}{\partial z^2}
   + z \frac{\partial R}{\partial z} + z^2 R = 0.
$$

```julia
z = Sym("z")
```

```julia
J0 = sympy.besselj(0, z)
```

```julia
d1J0 = J0.diff(z)
```

```julia
d2J0 = d1J0.diff(z)  
```

```julia
odeJ0 = z^2*d2J0 + z*d1J0 + z^2*J0
```

```julia
sympy.simplify(odeJ0)
```

The above calculation verifies that the Bessel function of the first kind satisfies the ordinary differential equation.  Let us do the same calculation for the second Bessel function.

```julia
Y0 = sympy.bessely(0, z)
```

```julia
d1Y0 = Y0.diff(z)
```

```julia
d2Y0 = d1Y0.diff(z)
```

```julia
odeY0 = z^2*d2Y0 + z*d1Y0 + z^2*Y0
```

```julia
sympy.simplify(odeY0)
```
