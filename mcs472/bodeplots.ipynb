# Bode Plots

Bode plots separate the radius and the angle in a polar plot.

### 1. polar plots

In polar coordinates, the unit circle has a very short representation:

$$
   r(\theta) = 1, \quad \theta \in [0, 2\pi[.
$$

```julia
r(theta) = 1
rangetheta = 0:0.01:2*pi
rt = [r(t) for t in rangetheta]
plot(rangetheta, rt, proj=:polar, label = "r(theta) = 1")
```

### 2. polar plot of the averaging filter

For the averaging filter, we derived the amplitude gain $r$ and phase shift $\phi$ to be

$$
    r = \frac{\sqrt{1 + \cos(\omega T)}}{\sqrt{2}}
$$

and

$$
    \phi = - \mbox{arctan} \left( \frac{\sin(\omega T)}{1 + \cos(\omega T)} \right).
$$

where $\omega = 2 \pi n$, $n$ is the frequency of the signal, and $T$ is the sampling rate.

The sampling rate $T = 1/1000$.

```julia
T = 1/1000
```

We let the frequencies range from 0 to 1000.  To make the plot, we evaluate the formulas for $r$ and $\phi$.

```julia
freq = 0:1:1000
omega = [2*pi*n for n in freq]
rw = [sqrt(1 + cos(w*T))/sqrt(2) for w in omega]
phi = [-atan(sin(w*T)/(1+cos(w*T))) for w in omega]
plot(phi, rw, proj=:polar, label="(1+z^(-1))/2")
```

### 3. the Bode plot for the averaging filter

In a Bode plot, the amplitude gain and phase shift are separately graphed against the input frequency.

```julia
plot(freq, rw, label="amplitude gain", legend=:right, xlabel="frequency Hz")
```

```julia
plot(freq, phi, label="phase shift", legend=:topright, xlabel="frequency Hz")
```

We can place the plots next to each other as follows.

```julia
gainplot = plot(freq, rw, label="gain", legend=:top, xlabel="frequency Hz")
phaseplot = plot(freq, phi, label="phase", legend=:topright, xlabel="frequency Hz")
plot(gainplot,phaseplot)
```
