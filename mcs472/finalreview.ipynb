# Taguchi quality control

Suppose we measure the quality of watches as the deviation
from the time at noon after one month without adjustments.

Watches which show a deviation of more than 3 minutes have
to be discarded at a cost of $\$10$.

We have to choose between two ways to produces watches:

1. Watches produced by process $A$ give after one month an average
   time of $12\!:\!01$ with a standard deviation of 1 minute.
    
2. Watches produced by process $B$ show $12\!:\!00$ on average,
   but with a standard deviation of 2 minutes.

Which process is best? $A$ or $B$?

1. Set up the quality loss function.
2. Use the expected loss to choose which process is best.

# Linear Programming

with the aim of maximizing the profit, subject to constraints:

1. We must produce at least 20 units of $A$, and
   offer at least $10$ units of $B$.

3. We have only 600 hours of labor available, 

   * one unit of $A$ requires two hours of labor,
   * while one unit of $B$ takes three hours of labor.

Do the following:

1. Set up the linear programming model for this problem.
2. Formulate the dual problem.

# Ordinary Differential Equations

there is already 20 ${\rm m}^3$ of water inside the ship.

   inside the ship as time advances.

   one can cover the crack in the hull and save the ship.
