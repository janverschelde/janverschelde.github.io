# Plotting fume profiles.

This notebooks makes some simple plots to represent a fume profile, assuming a Gaussian bell shaped curve.

```julia
using Plots
```

```julia
c0(x) = exp(-100*x^2)
```

```julia
plot(-2:0.01:2, c0, xlabel="x", label="c0(x)")
```

```julia
# savefig("figconcentration0.png")
```

```julia
ct(x) = exp(-x^2/4)/4
```

```julia
plot(-2:0.01:2, ct, xlabel="x", label="c(x, t)", ylims=[0,1])
```

```julia
# savefig("figconcentrationT.png")
```

A shifted profile...

```julia
c5(x) = exp(-100*(x - 5)^2)
```

```julia
plot(3:0.01:7, c5, xlabel="x", label="c0(x-5)")
```

```julia
# savefig("figconcentration5.png")
```

```julia
c5t(x) = exp(-(x-5)^2/4)/4
```

```julia
plot(3:0.01:7, c5t, xlabel="x", label="c(x-5t)", ylims=[0,1])
```

```julia
# savefig("figconcentration5T.png")
```
