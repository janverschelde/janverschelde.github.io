This notebook introduces the line segment intersection problem with the cgal-swig-bindings.

Consider three line segments defined by the tuples

((0, 0), (2, 2)), ((1, 0), (0, 2)), ((2, 0), (1, 1))

and compute their intersections.

```python
from CGAL.CGAL_Kernel import Point_2
from CGAL.CGAL_Kernel import Segment_2
from CGAL.CGAL_Kernel import intersection
```

# 1. Defining Points and Line Segments

The first segment has end points (0, 0) and (2, 2).  We define the points first before we define the first segment.

```python
a0 = Point_2(0, 0)
str(a0)
```

Other than the string representation of a point, we may retrieve the coordinates of the point separately via the ``x()`` and ``y()`` methods, as shown below.

```python
a0.x(), a0.y()
```

```python
b0 = Point_2(2, 2)
str(b0)
```

```python
s0 = Segment_2(a0, b0)
str(s0)
```

The end points of a line segment can be obtained via the ``vertex()`` method.

```python
str(s0.vertex(0)), str(s0.vertex(1))
```

# 2. Intersecting Line Segments

We define the first segment above in ``s0``.  For a proper intersection, we must define a second segment.
The second segment has end points (1, 0) and (0, 2), as defined below.

```python
a1 = Point_2(1, 0)
b1 = Point_2(0, 2)
s1 = Segment_2(a1, b1)
str(s1)
```

Now we are ready to apply the ``intersection()`` method of the ``CGAL_Kernel`` which returns an object we can query.  As the intersection of two line segments may be empty, we first apply the ``empty()`` method to the result of the intersection.

```python
ints0s1 = intersection(s0, s1)
ints0s1.empty()
```

The intersection of two line segments can be again a line segment.  To make sure we have a valid intersection point, we apply the ``is_Point_2()`` method to the result of the intersection.

```python
ints0s1.is_Point_2()
```

```python
ipts0s1 = ints0s1.get_Point_2()
str(ipts0s1)
```

Observe that we may obtain more accurate representations of the intersection point.

```python
ipts0s1.x(), ipts0s1.y()
```

```python
a2 = Point_2(2, 0)
b2 = Point_2(1, 2)
s2 = Segment_2(a2, b2)
str(s2)
```

```python
ints1s2 = intersection(s1, s2)
ints1s2.empty()
```

We see that ``s1`` and ``s2`` do not intersect.

# 3. Computing all intersection points

The code cell below reports all intersection points for our example.

```python
segments = []
segments.append(Segment_2(Point_2(0, 0), Point_2(2, 2)))
segments.append(Segment_2(Point_2(1, 0), Point_2(0, 2)))
segments.append(Segment_2(Point_2(2, 0), Point_2(1, 2)))
for i in range(len(segments)):
    for j in range(i+1, len(segments)):
        intsisj = intersection(segments[i], segments[j])
        print('segments', i, 'and', j, end=' ')
        if intsisj.empty():
            print('do not intersect')
        elif not intsisj.is_Point_2():
            print('do not intersect in a point')
        else:
            iptsisj = intsisj.get_Point_2()
            print('intersect :', iptsisj.x(), ',', iptsisj.y())
```

Of course, the running time of this algorithm is quadratic in the number of line segments.
