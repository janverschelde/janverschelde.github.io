This notebook illustrates the Delaunay triangulation from the spatial package of scipy.

```python
import numpy as np
```

```python
from scipy.spatial import Delaunay, delaunay_plot_2d
```

```python
from matplotlib import pyplot as plt
```

# The Delaunay triangulation of 10 random points

```python
points = np.random.rand(10, 2)
tri = Delaunay(points)
delaunay_plot_2d(tri)
plt.show()
```

Let us examine the attributes for the triangulation.

```python
print('the input points :')
print(tri.points)
```

```python
print('indices of the simplices :')
for simplex in tri.simplices:
    print(simplex)
```

```python
print('indices to neighbor simplices :')
for neighbor in tri.neighbors:
    print(neighbor)
```

```python
print('the equations :')
for equation in tri.equations:
    print(equation)
```

The equations are from the facets of the paraboloid construction used by qhull, see the documentation below.
