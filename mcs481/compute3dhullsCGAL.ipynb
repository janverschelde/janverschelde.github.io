# 1. a first example

The code in this section is copied from

`https://github.com/CGAL/cgal-swig-bindings/wiki/BindingsExamples#polyhedron_incremental_builderpy`

```python
from CGAL.CGAL_Polyhedron_3 import Polyhedron_modifier
from CGAL.CGAL_Polyhedron_3 import Polyhedron_3
from CGAL.CGAL_Polyhedron_3 import ABSOLUTE_INDEXING
from CGAL.CGAL_Kernel import Point_3
```

```python
#declare a modifier interfacing the incremental_builder
m = Polyhedron_modifier()
```

```python
#define a triangle
m.begin_surface(3,1)
m.add_vertex(Point_3(0,0,0))
m.add_vertex(Point_3(0,1,0))
m.add_vertex(Point_3(1,0.5,0))
m.begin_facet()
m.add_vertex_to_facet(0)
m.add_vertex_to_facet(1)
m.add_vertex_to_facet(2)
m.end_facet()
```

```python
m
```

```python
P = Polyhedron_3()
```

```python
#create the triangle in P
P.delegate(m)
print("(v,f,e) = ", P.size_of_vertices(), P.size_of_facets(), P.size_of_halfedges()/2)
```

```python
#clear the modifier
m.clear()
```

```python
#define another triangle, reusing vertices in the polyhedron
m.begin_surface(1,1,0,ABSOLUTE_INDEXING)
m.add_vertex(Point_3(-1,0.5,0))
m.begin_facet()
m.add_vertex_to_facet(1)
m.add_vertex_to_facet(0)
m.add_vertex_to_facet(3)
m.end_facet()
```

```python
#append a triangle incident to the existing one
P.delegate(m)
print("(v,f,e) = ", P.size_of_vertices(), P.size_of_facets(), P.size_of_halfedges()/2)
```

```python
assert P.is_valid()
```

# 2. The Wrapped Convex_hull_3

```python
from CGAL import CGAL_Convex_hull_3
```

Let us consider the points

{ (0,0,0), (3,0,0), (0,3,0), (3,3,0), (1,1,1), (2,1,1), (1,2,1), (2,2,1) }

```python
A = [Point_3(0,0,0), Point_3(3,0,0), Point_3(0,3,0), Point_3(3,3,0)]
```

```python
for p in A:
    print(p)
```

```python
B = [Point_3(1,1,1), Point_3(2,1,1), Point_3(1,2,1), Point_3(2,2,1)]
```

```python
L = A + B
for p in L:
    print(p)
```

```python
help(CGAL_Convex_hull_3.convex_hull_3)
```

```python
from CGAL.CGAL_Alpha_wrap_3 import Point_3_Vector
```

```python
pts = Point_3_Vector()
pts.reserve(8)
for point in L:
    pts.append(point)
```

```python
Q = Polyhedron_3()
```

```python
CGAL_Convex_hull_3.convex_hull_3(pts, Q)
```

```python
Q.size_of_facets()
```

```python
Q.size_of_vertices()
```

```python
for v in Q.points():
    print(v)
```

```python
for e in Q.edges():
    print('edge (', e.vertex().point(), end=' , ')
    print(e.next().vertex().point(), end=') \n')
```

```python
cnt = 0
for facet in Q.facets():
    print('facet', cnt, ':')
    e = facet.halfedge()
    print(e.vertex().point())
    print(e.next().vertex().point())
    print(e.next().next().vertex().point())
    cnt = cnt+1
```

Observe that the 12 facets are triangles.  From the picture on the lecture slides, we recall that there are 6 facets, where each facet has four vertices.  Each facet is thus represented by two triangles.
