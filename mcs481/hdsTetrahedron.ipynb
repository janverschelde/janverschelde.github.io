Half edge data structures are exported on three dimensional polyhedrons.  We consider a tetrahedron, as the convex hull of four points in 3-space.

```python
from CGAL.CGAL_Kernel import Point_3
from CGAL.CGAL_Polyhedron_3 import Polyhedron_3
```

```python
pt0 = Point_3(0, 0, 0)
pt1 = Point_3(1, 0, 0)
pt2 = Point_3(0, 1, 0)
pt3 = Point_3(0, 0, 1)
pts = [pt0, pt1, pt2, pt3]
print('Four points :')
for point in pts:
    print(point)
```

```python
tetrahedron = Polyhedron_3()
T = tetrahedron.make_tetrahedron(pt0, pt1, pt2, pt3)
T
```

```python
tetrahedron.is_valid()
```

```python
print('Number of vertices :', tetrahedron.size_of_vertices())
```

```python
V = tetrahedron.vertices()
for idx, vtx in enumerate(V):
    print('vertex', idx, ':', vtx.point())
```

```python
print('Number of half edges :', tetrahedron.size_of_halfedges())
```

```python
E = tetrahedron.edges()
for idx, edge in enumerate(E):
    print('edge', idx, '(', edge.vertex().point(), end=' , ')
    print(edge.next().vertex().point(), end=') ')
    print('has degree', edge.vertex_degree())
    print('   its predecessor : (', edge.prev().vertex().point(), end=' , ')
    print(edge.vertex().point(), ')')
    print('   incident facet has degree', edge.facet().facet_degree())
```

```python
print('Number of facets :', tetrahedron.size_of_facets())
```

```python
F = tetrahedron.facets()
for idx, facet in enumerate(F):
    print('facet', idx,'is triangle?', facet.is_triangle(), end=',')
    print(' with degree', facet.facet_degree())
    if facet.is_triangle():
        edge = facet.halfedge()
        print('traversing facet', idx, ':')
        print(edge.vertex().point())
        print(edge.next().vertex().point())
        print(edge.next().next().vertex().point())
        print(edge.next().next().next().vertex().point())
        print('traversing facet', idx, 'in the opposite order:')
        print(edge.vertex().point())
        print(edge.prev().vertex().point())
        print(edge.prev().prev().vertex().point())
        print(edge.prev().prev().prev().vertex().point())
```
