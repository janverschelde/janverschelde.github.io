Reading `https://doc.cgal.org/latest/Triangulation_2/` a triangulation of a set of points in the plane is obtained by adding *an infinite vertex*.  Imagine a pyramid that has the list of input points as its base.  Via this construction, all faces in the triangulation are triangles.

# 1. Triangulation of a Polygon

```python
from __future__ import print_function
from CGAL.CGAL_Kernel import Point_2
from CGAL.CGAL_Triangulation_2 import Triangulation_2
```

```python
L = []
L.append(Point_2(-2, 1))
L.append(Point_2(-2, 4))
L.append(Point_2(-1, 3))
L.append(Point_2(0, 5))
L.append(Point_2(0, 0))
L.append(Point_2(2, 1))
L.append(Point_2(2, 4))
L.append(Point_2(1, 3))
```

```python
T = Triangulation_2()
T.insert(L)
```

```python
T.dimension()
```

```python
T.number_of_faces()
```

```python
T.number_of_vertices()
```

```python
[str(p) for p in T.points()]
```

Let us look at all edges.

```python
cnt = 0
for e in T.all_edges():
    if not T.is_infinite(e):
        print(T.segment(e))
        cnt = cnt + 1
    else:
        print("infinite edge")
print('counted', cnt, 'finite edges')
```

If we look closely at the edges, then we observe that we obtained the triangulation of the convex hull of the list of points.

```python
cnt = 0
for f in T.all_faces():
    if not T.is_infinite(f):
        print(T.triangle(f))
        cnt = cnt + 1
    else:
        print("infinite face")
print('counted', cnt, 'finite faces')
```
