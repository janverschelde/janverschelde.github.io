We examine some special cases of Voronoi diagrams.

# The Bounding Box of a Voronoi Diagram

The code below makes
a plot of a Voronoi diagram of 10 points (k, noise[k]),
where the y-coordinate is randomly generated.
As the bounding box is computed based on the coordinates
of the sites, many vertices of the Voronoi diagram are
not shown, as they then lie outside the box.

```python
import numpy as np
from scipy.spatial import Voronoi, voronoi_plot_2d
from matplotlib import pyplot as plt
```

```python
noise = np.random.random(10)*1.75
line = [(k, noise[k]) for k in range(10)]
points = np.matrix(line)
vor = Voronoi(points)
voronoi_plot_2d(vor)
plt.show()
```

In the course we defined the bounding box as the box which contains all vertices of the Voronoi diagram, but this definition will lead to numerical problems when the coordinates of the vertices are too extreme with respect to the coordinates of the sites.  The above picture can be easily generated and shows the suboptimal solution of computing the bounding box based on the coordinates of the sites.

Changing the viewing axes of matplotlib could define a more favorable bounding box.

Perhaps even better, use a logarithmic scale for points outside the bounding box computed based on the coordinate of the sites.

# The Degree of a Vertex

For generic configurations of sites, the degree of a vertex is expected to be three.
But how high can the degree of a vertex be?

```python
nbr = 10
angle = 2*np.pi/nbr
circle = [(np.cos(k*angle), np.sin(k*angle)) for k in range(nbr)]
points = np.matrix(circle)
vor = Voronoi(points)
voronoi_plot_2d(vor)
plt.show()
```

# A Cluster of Vertices

What if we add some noise to the points?

```python
noise = np.random.random(nbr)*0.2
circle = [(np.cos(k*angle), np.sin(k*angle)+noise[k]) for k in range(nbr)]
points = np.matrix(circle)
vor = Voronoi(points)
voronoi_plot_2d(vor)
plt.show()
```
