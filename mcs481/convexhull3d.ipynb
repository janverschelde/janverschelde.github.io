# 1. Three Dimensional Convex Hulls

Let us generate 20 random points in 3-space.

```python
import numpy as np
from scipy.spatial import ConvexHull
```

```python
points = np.random.rand(20, 3)
print('20 random points :')
print(points)
```

```python
hull = ConvexHull(points)
```

```python
print('the vertices :')
print(hull.vertices)
print('the facets :')
for simplex in hull.simplices:
    print(simplex)
```

```python
n = len(hull.vertices)
n
```

```python
m = len(hull.simplices)
m
```

```python
2*n - 4
```

For a random point collection, we expect all facets to be triangles.  The facets lie in the planes, reported in the `equations` attribute of the computed hull.

```python
hull.equations
```
