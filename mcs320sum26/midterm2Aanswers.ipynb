# Question 5

Write an efficient recursive function to compute
$$
         q_0(x) = 1, ~~ q_1(x) = x + 1, ~~ \mbox{for } n > 1:
         q_n(x) = (2 x - 1) q_{n-1}(x) + (x + 1) q_{n-2}(x).
$$
Make sure your function can compute $q_{100}(x)$.

Do *NOT* use ``cached_function``.

## answer to question 5

```python
reset()
```

```python
def q(n, x, D={}):
    """
    Returns p(x) using memoization.
    """
    if (n, x) in D:
        return D[(n, x)]
    else:
        if n == 0:
            result = 1
        elif n == 1:
            result = x + 1
        else:
            result = expand((2*x-1)*q(n-1, x) + (x+1)*q(n-2, x))
        D[(n,x)] = result
        return result
```

```python
[q(k, x) for k in range(5)]
```

```python
q(100, x)
```

# Question 8

Consider $x(t)$ a function of $t$ as the solution of the
      initial value problem
$$
   \frac{d^2}{d~\!t^2}~\!x(t) - \frac{d}{d~\!t}~\!x(t) - 2 x(t) = 0,
   \quad x(0) = 1, \quad x'(0) = 0.
$$
Compute $x(t)$ using the Laplace transform.

What is the algebraic equation that must be solved?

## answer to question 8

```python
reset()
```

```python
t = var('t')
```

```python
x = function('x')(t)
```

```python
ode = diff(x, t, 2) - diff(x, t) - 2*x
show(ode)
```

```python
s = var('s')
Lode = ode.laplace('t', 's')
show(Lode)
```

```python
T = var('T')
Teq = Lode.subs({laplace(x, t, s): T})
show(Teq)
```

The above equation shows the algebraic equation in the Laplace tranform ``T`` with the initial conditions which has to be solved.

```python
sol = solve(Teq, T)
show(sol)
```

```python
init = {x(t=0): 1, diff(x,t).subs(t=0): 0}
show(init)
```

```python
Ts = sol[0].rhs().subs(init)
ft = inverse_laplace(Ts, s, t)
show(ft)
```
