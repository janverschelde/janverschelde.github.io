# Question 1

Compute the first 10 rational approximations of $\sin(\sqrt{\pi})$
with increasing size of denominator,
starting at a denominator bounded by 10 and
going up to a bound of $10^{10}$.

Verify the accuracy of the computed rational approximations
via their relative errors.

## answer to question 1

```python
s = sin(sqrt(pi))
show(s)
```

```python
q = [s.n(digits=20).nearby_rational(max_denominator=10^k) for k in range(1,11)]
q
```

```python
e = [abs(x - s.n(digits=20))/abs(x) for x in q]
e
```

```python
dp = [int(abs(log(x, 10))) for x in e]
dp
```

The list ``dp`` contains the number of correct decimal places in the rational approximations.

# Question 6

Let

``s = lambda n: float(sum([exp(-(k/n)**2)*cos(3*k/n) for k in range(n)])/n)``

be a function to approximate 
$\displaystyle \int_{0}^{1} \exp(-x^2) \cos(3x) dx$.


1. Define a Python function ``p`` which does the same as ``s``.
   Explain why ``p`` is inefficient.

   For correctness, compare the output of ``p(1000)`` and ``s(1000)``.

2. Apply Cython to improve the efficiency of ``p``.  Verify the correctness.

   Time the execution of the Cython function for $n = 10000$, compare with timings of ``p``.

## answer to question 6

```python
reset()
```

```python
s = lambda n: float(sum([exp(-(k/n)**2)*cos(3*k/n) for k in range(n)])/n)
```

```python
def p(n):
    return float(sum([exp(-(k/n)**2)*cos(3*k/n) for k in range(n)])/n)
```

```python
s1000 = s(1000)
s1000
```

```python
p1000 = p(1000)
p1000
```

```python
s1000 - p1000
```

The function ``p`` is inefficient because all arguments of ``exp()`` and ``cos()`` are rational numbers ``k/n`` which are kept in their exact form.  Only at the very end, after representing the sum as a large expression, does the sum get evaluated in floating-point arithmetic.

```python
%%cython
cdef extern from "math.h":
    double cos(double)

cdef extern from "math.h":
    double exp(double)

def c(long n):
    return sum([exp(-(float(k)/n)**2)*cos(3*float(k)/n) for k in range(n)])/n
```

```python
c1000 = c(1000)
c1000
```

```python
abs(s1000 - c1000)
```

We see that the difference between the output of the original ``s`` and its Cythonized version is of the order double precision.

```python
timeit('s(10000)')
```

```python
timeit('c(10000)')
```

Observe the drop in magnitude in the times, from milliseconds to microseconds.

# Question 9

Consider the expressions $p = x + 4$ and
$\displaystyle q = \frac{x^2 + x - 12}{x-3}$.

1. Are the expressions *symbolically* the same?

2. Are $p$ and $q$ *numerically* equivalent?

Justify your answers by the appropriate computations.

## answer to question 9

```python
reset()
```

```python
p = x+4
show(p)
```

```python
q = (x^2 + x - 12)/(x - 3)
show(q)
```

1. As symbolic expressions, $p$ and $q$ are not the same, but if we normalize $q$ by removing the common factor from numerator and denominator, then we arrive at the same expression.

```python
nq = q.numerator(normalize=True)
show(nq)
```

```python
p - nq
```

2. The numerical equality test happens via comparing the results of the evaluation at a random number.

```python
r = CC.random_element()
r
```

```python
vp = p(x=r)
show(vp)
```

```python
vq = q(x=r)
show(vq)
```

```python
vp - vq
```

We see that the difference between their values at a random number is of a magnitude equal to the double precision.  So, numerically, $p$ and $q$ are equivalent, except of course close to the special value where the denominator of $q$ becomes zero.
