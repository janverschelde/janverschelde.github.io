For some parameter $s$,
consider the space curve $(x(s,t), y(s,t), z(s,t))$
with its coordinates defined by
$$
   x(s,t) = r(s, t) \cos(4t),  \quad
   y(s,t) = r(s, t) \sin(4t),  \quad
   z(s,t) = r(s, t) \sin(s t),
$$
where $r(s,t) = 2 + 4/5 \cos(s t)$, for $t$ ranging from 0 to $2\pi$.

Write code to define an animation with eleven frames.

The frames are defined by the value for $s$, which ranges from 1 to 2.

# answer

```python
r(s,t) = 2 + (4/5)*cos(s*t)
x(s,t) = r(s,t)*cos(4*t)
y(s,t) = r(s,t)*sin(4*t)
z(s,t) = r(s,t)*sin(s*t)
curve(s) = (x, y, z)
show(curve)
```

```python
frame1 = parametric_plot3d(curve(1), (t, 0, 2*pi), thickness=3)
show(frame1)
```

```python
frame10 = parametric_plot3d(curve(2), (t, 0, 2*pi), thickness=3)
show(frame10)
```

```python
frames = [parametric_plot3d(curve(k/10), (t, 0, 2*pi), thickness=3) for k in range(10,21)]
```

```python
a = animate(frames, xmin=-3, xmax=3, ymin=-3, ymax=3, zmin=-3, zmax=3)
```
