# Question 1

Define a function ``P`` which returns
$p_i(x_1, x_2, \ldots, x_n) = x_i x_1 - x_{i+1}$,
for $i=1,2, \ldots, n-1$.

The input to ``P`` is the index $i$ and the list 
$x_1, x_2, \ldots, x_n$.

Demonstrate that your function works for $n = 8$.

## answer to question 1

```python
X = [var('x' + str(k)) for k in range(1, 9)]
show(X)
```

```python
p = lambda i, X: X[i-1]*X[0] - X[i]
```

```python
[p(k, X) for k in range(1, 8)]
```

# Question 2

Let $P = (2, 3)$ be the point on the curve
$x^2 y^3 - 5 x^3 - 7y^2 - 5 = 0$.

Compute the tangent line to the curve at $P$.  
What is the slope of this tangent line?

## answer to question 2

```python
reset()
```

```python
x, y = var('x, y')
curve = x^2*y^3 - 5*x^3 - 7*y^2 - 5
show(curve)
```

Using implicit differentiation, we declare Y as a function of x.

```python
Y = function('Y')(x)
Ycurve = curve.subs({y: Y})
show(Ycurve)
```

```python
dcurve = diff(Ycurve, x)
show(dcurve)
```

```python
dy = diff(Y, x)
sy = solve(dcurve, dy)
show(sy)
```

```python
slope = sy[0].rhs().subs({x:2, Y(x=2):3})
slope
```

```python
tangent = y - 3 - slope*(x-2)
show(tangent)
```

An alternative way is to compute a Taylor series, truncated after degree 1.

```python
t = taylor(curve, (x, 2), (y, 3), 1)
show(t)
```

```python
show(t/66)
```

```python
slope = - t.coefficient(x, 1)/t.coefficient(y, 1)
slope
```
