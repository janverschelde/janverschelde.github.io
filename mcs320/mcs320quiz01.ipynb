Consider $\displaystyle s = \sin\left(\sqrt{13}\right)$.

1. Compute a rational approximation for $s$, accurate to 3 decimal places.

2. Verify that the accuracy of your rational approximation is 3 decimal places.

   Write an explanation for your verification.

3. Compute the sequence of the first ten consecutive rational approximations for $s$,
   accurate respectively with 1, 2, up to 10 decimal places (10 included).

# Answer to Part 1

```python
s = sin(sqrt(13))
show(s)
```

```python
a = QQ(s.n(digits=3))
show(a)
```

# Answer to Part 2

```python
absolute_error = abs(s-a).n(digits=4)
show(absolute_error)
```

```python
relative_error = (absolute_error/abs(s)).n(digits=4)
show(relative_error)
```

```python
'%.2e' % relative_error
```

The relative error is $1.45 \times 10^{-4}$.  This verifies that the approximation has indeed the first 3 decimal places correct.

# Answer to Part 3

```python
L = [QQ(s.n(digits=k)) for k in range(1, 11)]
show(L)
```
