# Question 1

Use the first 20 convergents of the continued fraction of $\sqrt{\pi/3}$
to define the first 20 rational approximations of $\sqrt{\pi/3}$.

Compute the relative errors of
those 20 rational approximations to verify the accuracy.

## answer to question 1

```python
s = sqrt(pi/3)
show(s)
```

```python
c = continued_fraction(s)
c
```

```python
q = list(c.convergents()[:20])
q
```

Let us now compute the relative errors.

```python
e = [(abs(a-s)/abs(s)).n(digits=30) for a in q]
e
```

Observe that we need to evaluate the relative errors with enough precision.

To see the number of correct decimal places, we format the errors in scientific notation.

```python
['%.2e' % x for x in e]
```

# Question 3

Explain the difference between sequential and simultaneous substitution. 

Illustrate the difference with an example from the course.

## answer question 3

```python
reset()
```

In a sequential substitution, the symbols in an expression are replaced one after the other, whereas in a simultaneous substitution, all symbols are replaced at the same time.

```python
a, b, c = var('a, b, c')
e = a + 2*b + 3*c
show(e)
```

A sequential substitution replaces ``a`` by ``b``, then ``b`` by ``c``, and then ``c`` by ``a`` as done below.

```python
e.subs({a:b}).subs({b:c}).subs({c:a})
```

A simultaneous substitution is done by one ``subs`` as done below.

```python
e.subs({a:b, b:c, c:a})
```

# Question 4

Consider the function 
$\displaystyle f(n) = \frac{2}{n} \sum_{k=1}^{n-1}
\left( \frac{k}{n} \right)^2 \cos\left( \frac{k}{n} \right)$.

1. Code the definition of $f$ into a list comprehension
   to make a function ``F``
   which for some positive integer $n$ returns the floating-point 
   approximation of $f(n)$ in double precision.

2. Time the execution of ``F`` for $n=10,000$.
      
   Explain why ``F`` is inefficient.

3. Use Cython to define the function ``C``
   which returns the floating-point value of ``F``$(n)$.

   Verify that ``C`` is correct.

   For $n=10,000$, compare the running time of ``C`` with ``F``.

## answer to question 4

```python
reset()
```

```python
def F(n):
    return float(2*sum([(k/n)^2*cos(k/n) for k in range(1,n)])/n)
```

```python
timeit('F(10000)')
```

The Python function ``F`` is inefficient because the sum is evaluated symbolically and the reduction to a ``float`` is done only at the very end.

After adding the decorator ``%%cython`` we import the ``cos`` from the C math library, add type declarations for ``n``, ``k``, replace the ``^`` by ``**`` and divide by ``float(n)`` to force the floating-point division.

```python
%%cython
cdef extern from "math.h":
    double cos(double)

def C(long n):
    cdef long k
    return 2*sum([(k/float(n))**2*cos(k/float(n)) for k in range(1,n)])/n
```

```python
F(100)
```

```python
C(100)
```

```python
abs(F(100) - C(100))
```

The difference in values returned by ``F`` and ``C`` is of the magnitude of double precision.

```python
timeit('C(10000)')
```

Indeed, we observe that the execution time has dropped from milliseconds to microseconds.
