# Question 6

Let

``s = lambda n: float(sum([exp(-(k/n)**2)*cos(3*k/n) for k in range(n)])/n)``

be a function to approximate 
$\displaystyle \int_{0}^{1} \exp(-x^2) \cos(3x) dx$.


1. Define a Python function ``p`` which does the same as ``s``.
   Explain why ``p`` is inefficient.

   For correctness, compare the output of ``p(1000)`` and ``s(1000)``.

2. Apply Cython to improve the efficiency of ``p``.  Verify the correctness.

   Time the execution of the Cython function for $n = 10000$, compare with timings of ``p``.
