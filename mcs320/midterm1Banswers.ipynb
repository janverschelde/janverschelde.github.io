# Question 1

Compute the first 10 consecutive rational approximations of $\sin(\pi^2)$
with the consecutive convergents of the continued fraction approximation.

Verify the accuracy of the computed rational approximations
via their relative errors.

## answer to question 1

```python
s = sin(pi^2)
show(s)
```

```python
cfs = continued_fraction(s)
cfs
```

```python
cvg = cfs.convergents()
cvg
```

```python
q = [cvg[k] for k in range(10)]
show(q)
```

```python
e = [abs(x - s.n(digits=20))/abs(s.n(digits=20)) for x in q]
e
```

```python
L = [abs(log(x, 10)) for x in e]
L
```

```python
dp = [int(x) for x in L]
dp
```

The numbers in the list ``dp`` show the correct decimal places for the corresponding rational approximations in the list ``q``.

of the expanded form with the coefficients of $p$.

# Question 4

Explain the importance of normalization in symbolic computation.

Illustrate with good examples the difference between a symbolic
and a numeric test on the equality of two expressions.

## answer to question 4

Normalization is used to decide whether two expressions are the same, which is important as many equivalent mathematical objects may have many different expressions.

```python
reset()
```

```python
x, y = var('x, y')
```

```python
e1 = x*(x + y)
show(e1)
```

```python
e2 = x^2 + x*y
show(e2)
```

Mathematically, both ``e1`` and ``e2`` are equivalent, but they are different expressions.  A normalization occurs by fully expanding.

```python
ee1 = expand(e1)
show(ee1)
```

```python
ee1 - e2
```

A numerical test on equality of two expressions compares the values of the expressions evaluated at random points.

```python
xv = CC.random_element()
yv = CC.random_element()
e1(x=xv, y=yv) - e2(x=xv, y=yv)
```

We observe that the error is of a magnitude equal to double precision.

# Question 6

Let 

``s = lambda n: float(sum([exp(-(k/n)**2)*cos(3*k/n) for k in range(n)])/n)``

be a function to approximate 
$\displaystyle \int_{0}^{1} \exp(-x^2) \cos(3x) dx$.

1. Time the execution of ``s`` for $n = 10000$.  Explain why ``s`` is inefficient.

2. Apply vectorization to improve the efficiency.  Verify the correctness.

   Time the execution of the vectorized function for $n = 10000$,
   compare with timings of ``s``.

## answer to question 6

```python
reset()
```

```python
s = lambda n: float(sum([exp(-(k/n)**2)*cos(3*k/n) for k in range(n)])/n)
```

```python
timeit('s(10000)')
```

The function ``s`` is inefficient because the conversion to ``float`` happens only at the very end, whereas all 10000 ``k/n`` are computed exactly.

```python
def vs(n):
    from numpy import linspace, exp, cos, sum
    x = linspace(0, 1-1.0/n, n)
    return sum(exp(-x**2)*cos(3*x))/n
```

```python
s100 = s(100)
s100
```

```python
v100 = vs(100)
v100
```

```python
abs(s100 - v100)
```

The error is of a magnitude equal to the double precision.

```python
timeit('vs(10000)')
```

The speedup of the vectorized version over the original function is significant: we go from milliseconds to microseconds.

Let $\displaystyle q = \frac{3.5 x^7 - 1}{x^7 - 1}$ be
a polynomial with complex coefficients.

1. Compute the partial fraction decomposition of $q$.
   Relate the denominators of the partial fraction decomposition
   to the roots of the denominator of $q$.

2. Given the partial fraction decomposition of $q$, reconstruct $q$.
   Describe what you observe when you normalize.

## answer to question 7

```python
reset()
```

```python
x = polygen(CC)
q = (3.5*x^7 - 1)/(x^7 - 1)
show(q)
```

```python
dq = q.partial_fraction_decomposition()
show(dq)
```

The partial fraction decomposition allows to rewrite $q$ as a sum of rational expressions, where the roots of the denominator of $q$ appear in the denominators of the rational expressions in the sum.

```python
eq = dq[0] + sum(dq[1])
show(eq)
```

In the reconstructed partial fraction decomposition, we see tiny coefficients of the order of double precision appearing at places where the coefficients were zero $q$.

# Question 9

Transform $\displaystyle q = (a-b)^2 + \frac{(a+b)^3}{(a-b)^4}$
into $\displaystyle r = \frac{(a-b)^6 + (a+b)^3}{(a-b)^4}$,
without typing $r$.

## answer to question 9

```python
reset()
```

```python
a, b = var('a, b')
q = (a-b)^2 + (a+b)^3/(a-b)^4
show(q)
```

```python
u, v = var('u, v')
uvq = q.subs({a-b: u, a+b: v})
show(uvq)
```

```python
fq = uvq.numerator()/uvq.denominator()
show(fq)
```

```python
r = fq(u=a-b, v=a+b)
show(r)
```
