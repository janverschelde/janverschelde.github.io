# Question 1

Let $S = \sqrt{\pi/3}$.

1. Compute a sequence of nearby rational approximations for $S$

   where the size of the denominator of the $k$-th approximation is bounded by $10^k$.

2. Compute the relative errors of those 20 rational approximations to verify the accuracy.

## answer to question 1

```python
S = sqrt(pi/3)
show(S)
```

Observe that we have to evaluate $\sqrt{pi/3}$ to a high enough precision.

```python
r = S.n(digits=64)
r
```

```python
q = [r.nearby_rational(max_denominator=10^k) for k in range(1, 21)]
q
```

The computation of the relative errors is below.

```python
e = [(abs(a-S)/abs(S)).n(digits=64) for a in q]
e
```

Observe that also the relative errors were computed in a high precision.

To see the number of correct decimal places, we convert to scientific notation.

```python
['%.2e' % x for x in e]
```

# Question 2

Consider $p = x^4 + 3 x^3 + 2 x^2 + 2 x + 4$ as a polynomial with
coefficients in $\mathbb Q$.

Add sufficiently many formal roots to $\mathbb Q$ so that $p$
can be written as the product of four linear factors.

What are the roots in the extended field?

## answer to question 2

```python
reset()
```

```python
x = polygen(QQ)
p = x^4 + 3*x^3 + 2*x^2 + 2*x + 4
show(p)
```

```python
p.is_irreducible()
```

Because ``p`` is not irreducible, we must select an irreducible factor to define a field extension.

```python
f = factor(p)
f
```

```python
F = list(f)
F
```

```python
f2 = F[1][0]
show(f2)
```

```python
K.<a> = QQ.extension(f2)
K
```

```python
p2 = K[x](p)
fp2 = factor(p2)
fp2
```

Because the last factor is still not linear, we must extend once more.

```python
Fp2 = list(fp2)
Fp2
```

```python
f3 = Fp2[2][0]
f3
```

```python
K2.<b> = K.extension(f3)
K2
```

```python
q = K2[x](p)
factor(q)
```

After casting ``p`` into this newest field ``K2`` we see that ``q`` factors and its four roots are in the list of the output cell below.

```python
q.roots(K2)
```

# Question 4

Consider the function 
$\displaystyle f(n) = \frac{2}{n} \sum_{k=1}^{n-1}
\left( \frac{k}{n} \right)^2 \cos\left( \frac{k}{n} \right)$.

1. Code the definition of $f$ into a list comprehension
   to make a function ``F``
   which for some positive integer $n$ returns the floating-point 
   approximation of $f(n)$ in double precision.

2. Time the execution of ``F`` for $n=10,000$.
      
   Explain why ``F`` is inefficient.

3. Use numpy to vectorize ``F``, as defined in the function ``V``$(n)$.

   Verify that ``V`` is correct.

   For $n=10,000$, compare the running time of ``V`` with ``F``.

## answer to question 4

```python
reset()
```

```python
def F(n):
    return float(2*sum([(k/n)^2*cos(k/n) for k in range(1,n)])/n)
```

```python
timeit('F(10000)')
```

The Python function ``F`` is inefficient because the sum is evaluated symbolically and the reduction to a ``float`` is done only at the very end.

```python
def V(n):
    from numpy import sum, cos, arange
    x = arange(1, n)/n
    return 2*sum(x^2*cos(x))/n
```

Let us verify the correctness of ``V`` by comparing its value at $n = 100$ at ``F``.

```python
F(100)
```

```python
V(100)
```

```python
abs(F(100) - V(100))
```

Observe that the difference between ``V`` and ``F`` is about the double precision.

```python
timeit('V(10000)')
```

Indeed, we observe that the execution of the vectorized function runs a whole lot faster than the original Python function: dropped from milliseconds to microseconds.

# Question 5

Make a fast callable object of
$\displaystyle x - \frac{y^2}{2x + y^2 - 3}$
and draw its expression tree.

## answer to question 5

```python
reset()
```

```python
from sage.ext.fast_callable import ExpressionTreeBuilder
etb = ExpressionTreeBuilder(vars=['x','y'])
x = etb.var('x')
y = etb.var('y')
e = x - y^2/(2*x + y^2 - 3)
e
```

Fast callable expressions have a binary expression tree, which we start to define at the leaves.

```python
Lx = LabelledBinaryTree([], 'x')
Ly = LabelledBinaryTree([], 'y')
L2 = LabelledBinaryTree([], '2')
L3 = LabelledBinaryTree([], '3')
```

```python
Ly2 = LabelledBinaryTree([Ly, L2], 'ipow')
ascii_art(Ly2)
```

```python
L2x = LabelledBinaryTree([L2, Lx], 'mul')
ascii_art(L2x)
```

```python
L2xy2 = LabelledBinaryTree([L2x, Ly2], 'add')
ascii_art(L2xy2)
```

```python
num = LabelledBinaryTree([L2xy2, L3], 'sub')
ascii_art(num)
```

```python
trm2 = LabelledBinaryTree([Ly2, num], 'div')
ascii_art(trm2)
```

```python
tree = LabelledBinaryTree([Lx, trm2], 'sub')
ascii_art(tree)
```

# Question 6

Let $\displaystyle q = \frac{4.12 x^6 - 1}{x^6 - 1}$ be
a quotient of polynomials with complex coefficients.

1. Compute the partial fraction decomposition $p$ of $q$.
   Relate the denominators of the partial fraction decomposition $p$
   to the roots of the denominator of $q$.

2. Given the partial fraction decomposition $p$ of $q$, reconstruct $q$.
   
   Describe what you observe when you normalize.

## answer to question 6

```python
reset()
```

```python
x = polygen(CC)
q = (4.12*x^6 - 1)/(x^6 - 1)
show(q)
```

```python
dq = q.partial_fraction_decomposition()
show(dq)
```

The partial fraction decomposition allows to rewrite $q$ as a sum of rational expressions, where the roots of the denominator of $q$ appear in the denominators of the rational expressions in the sum.

```python
q.denominator().roots(CC)
```

To reconstruct ``q`` from ``p`` we proceed as follows.

```python
eq = dq[0] + sum(dq[1])
show(eq)
```

In the reconstructed partial fraction decomposition, we see tiny coefficients of the order of double precision appearing at places where the coefficients were zero $q$.

# Question 7

Transform $p = (2 y - z)^6 - x^4$ 

into $r = ((2 y - z)^3 + x^2)((2 y - z)^3 - x^2)$,
*without* typing $r$.

## answer to question 7

```python
reset()
```

```python
x, y, z = var('x, y, z')
p = (2*y - z)^6 - x^4
show(p)
```

To shield the ``2*y - z`` from expanding, we replace ``2*y - z`` by ``u``.

```python
u = var('u')
q = p.subs({2*y - z: u})
show(q)
```

```python
f = factor(q)
show(f)
```

```python
r = f(u = z - 2*y)
show(r)
```
