## Interval Arithmetic for Accurate Results

The goal of this project is to apply interval arithmetic
to evaluate expressions accurately:

1. Evaluating expressions at an interval returns an interval.
   The center of the interval approximates the value of expression.
   The diameter of the interval on return gives an upper bound on
   the error of the computed value.

2. Given a desired accuracy, we increase the precision of the
   interval arithmetic gradually till the upper bound of 
   the error is sufficiently small.

Below we illustrate an interval arithmetic computation with 8 bits of precision.

```python
RIF8 = RealIntervalField(prec=8)
RIF8
```

The number 1/10 has no finite binary expansion.

```python
onetenth = RIF8(1/10)
onetenth
```

```python
onetenth.str(style='brackets')
```

```python
onetenth.diameter()
```

Because 1/10 has no finite binary representation, in a number system with 8 bits of precision, we have an error of $5 \times 10^{-3}$.

```python
ten = 1/onetenth
ten.str(style='brackets')
```

```python
ten
```

```python
ten.center()
```

```python
ten.diameter()
```

With 8 bits of precision, we evaluated 1/(1/10) and obtained an error of $1.3 \times 10^{-2}$.

If we are satisfied with 2 decimal places of accuracy, then 8-bit floating-point
arithmetic will do.  Otherwise, for a given desired accuracy, we gradually double
the precision under the upper bound for the error is sufficiently small.

Know the distinction between ``1/RIF8(1/10)`` and ``RIF8(1/(1/10))``.

```python
1/RIF8(1/10) 
```

```python
RIF8(1/(1/10))
```

The second result is exact, while the first one is not.

# 1. The Accuracy of Evaluating an Expression

Consider the expression for $X$ below
$$
   X = \left(
	  \frac{27/10 - e}{\pi - (\sqrt{2} + \sqrt{3})}
       \right)^{67/16}.
$$

## Assignment One.

1. Evaluate the expression for $X$ with 16-bit precision interval arithmetic.
   What is the upper bound on the error?

2. How many bits of precision will be required if we want to know the value of $X$
   with 8 decimal places of accuracy?
   Gradually double the precision,
   stopping when the upper bound on the error is less than $5 \times 10^{-8}$.
   Report the bound on the error in each step.

3. Compare the centers of the intervals of the last two approximations.
   How many decimal places are in agreement?

### answer to the first part of assignment one

```python
reset()
```

```python
RIF16 = RealIntervalField(prec=16)
```

```python
Xnum = RIF16(27/10) - RIF16(e)
Xnum
```

```python
Xden = RIF16(pi) - (sqrt(RIF16(2)) + sqrt(RIF16(3)))
Xden
```

```python
Xfrac = Xnum/Xden
Xfrac
```

```python
Xexp = RIF16(67/16)
Xexp
```

```python
X = Xfrac^Xexp
X
```

```python
X.diameter()
```

As the error is $2 \times 10^{-1}$, we have only one decimal place of accuracy.

### answer to the second part of assignment one

Let us double the precision, from 16 to 32 bits.

```python
RIF32 = RealIntervalField(prec=32)
Xnum = RIF32(27/10) - RIF32(e)
Xden = RIF32(pi) - (sqrt(RIF32(2)) + sqrt(RIF32(3)))
Xfrac = Xnum/Xden
Xexp = RIF32(67/16)
X32 = Xfrac^Xexp
X32.diameter()
```

We see that the error has dropped to $3 \times 10^{-6}$, close, but not yet at 8 decimal places.

We double again, from 32 to 64 bits.

```python
RIF64 = RealIntervalField(prec=64)
Xnum = RIF64(27/10) - RIF64(e)
Xden = RIF64(pi) - (sqrt(RIF64(2)) + sqrt(RIF64(3)))
Xfrac = Xnum/Xden
Xexp = RIF64(67/16)
X64 = Xfrac^Xexp
X64.diameter()
```

Now we see that the bound on the error is at $7 \times 10^{-16}$.

### answer to the third part of assignment one

If we compare the centers of ``X32`` and ``X64``:

```python
show(X32.center())
show(X64.center())
```

We observe that the first 8 decimal places of the 32-bit approximation are identical.

# 2. The Area of a Thin Triangle

To compute the area $A$ of a triangle with sides $a$, $b$, and $c$,
we evaluate

$$ 
   a = 7;  b = 7/2 + 3 \times 2^{-111};  c = b
$$

$$ 
   s = (a + b + c)/2;  A = \sqrt{s(s-a)(s-b)(s-c)}
$$

## Assignment Two.

1. Evaluate the formulas with 8-bit precision interval arithmetic.

   What is the upper bound for the error on $s$?  How accurate is $A$?

2. If we want the area $A$ to be accurate with 8 decimal places,
   then how many bits in the precision of the interval arithmetic are needed?

   Gradually double the precision, stopping when the upper bound on the error
   is less than $5 \times 10^{-8}$.

   What is the relation between the bound on the error of $s$ and the error on $A$?

3. Compare the centers of the last two approximations and
   report the number of decimal places that agree between those centers.

   If the accuracy of the approximation for $A$ is less than 8 decimal places,
   then continue doubling the precision till it is.

### answer to the first part of assignment two

```python
RIF8 = RealIntervalField(prec=8)
RIF8
```

```python
a = RIF8(7);  b = RIF8(7/2) + 3*RIF8(2^(-111));  c = b
s = (a+b+c)/2
s
```

```python
s.diameter()
```

```python
A = sqrt(s*(s-a)*(s-b)*(s-c))
A
```

```python
A.diameter()
```

While the upper bound on the error on $s$ is $9 \times 10^{-3}$, we have no accuracy for $A$, as the upper bound on the error is 2.4.

### answer to the second part of assignment two

```python
RIF16 = RealIntervalField(prec=16)
a = RIF16(7);  b = RIF16(7/2) + 3*RIF16(2^(-111));  c = b
s = (a+b+c)/2
print('error on s :', '%.1e' % s.diameter())
A = sqrt(s*(s-a)*(s-b)*(s-c))
print('error on A :', '%.1e' % A.diameter())
```

```python
RIF32 = RealIntervalField(prec=32)
a = RIF32(7);  b = RIF32(7/2) + 3*RIF32(2^(-111));  c = b
s32 = (a+b+c)/2
print('error on s :', '%.1e' % s32.diameter())
A32 = sqrt(s32*(s32-a)*(s32-b)*(s32-c))
print('error on A :', '%.1e' % A32.diameter())
```

The upper bound on $A$ is $6 \times 10^{-4}$, which is not enough.  We observe the wide gap between the erron on $s$, which is $5 \times 10^{-10}$, and the error on $A$.  As 32-bit interval arithmetic does not suffice, we double the precision, from 32 to 64.

```python
RIF64 = RealIntervalField(prec=64)
a = RIF64(7);  b = RIF64(7/2) + 3*RIF64(2^(-111));  c = b
s64 = (a+b+c)/2
print('error on s :', '%.1e' % s64.diameter())
A64 = sqrt(s64*(s64-a)*(s64-b)*(s64-c))
print('error on A :', '%.1e' % A64.diameter())
```

With 64-bit interval arithmetic, we have an upper bound on the error of $8 \times 10^{-9}$, which is less than the required $5 \times 10^{-8}$.  We observe the widening of the big gap between the error on $s$ and the error on $A$.

### answer to part three of assignment two

```python
show(A32.center())
show(A64.center())
```

Oops, how do we explain this, other than that the value for $A$ is much too small.

```python
RIF128 = RealIntervalField(prec=128)
a = RIF128(7);  b = RIF128(7/2) + 3*RIF128(2^(-111));  c = b
s128 = (a+b+c)/2
print('error on s :', '%.1e' % s128.diameter())
A128 = sqrt(s128*(s128-a)*(s128-b)*(s128-c))
print('error on A :', '%.1e' % A128.diameter())
```

# 3. The Quadratic Formula

Consider the application of the quadratic formula to compute
the roots $r_1$ and $r_2$ of $3 x^2 + 100 x + 2$, evaluating
$$
   a = 3;  b = 100;  c = 2;  d = \sqrt{b^2 - 4 ac} 
$$
$$
   r_1 = (-b + d)/(2a);  r_2 = (-b - d)/(2a)
$$

## Assignment Three. 

1. Evaluate the formulas for the discriminant $d$ and for $r_1$, $r_2$
   with 8-bit precision interval arithmetic.

   Report the upper bound for the error on $d$, $r_1$, and $r_2$.

   Can you predict the upper bound for the error on the roots via the error on $d$?

2. If we want the roots to be accurate with 8 decimal places,
   then how many bits in the precision of the interval arithmetic are needed?

   Gradually double the precision, stopping when the upper bound on the error
   is less than $5 \times 10^{-8}$.

   What is the relation between the bound on the error of $d$ and the largest error on the roots?

3. Compare the centers of the last two approximations and
   report the number of decimal places that agree between those centers.

### answer to part one of assignment three

```python
reset()
```

```python
RIF8 = RealIntervalField(prec=8)
RIF8
```

```python
a = RIF8(3); b = RIF8(100); c = RIF8(2)
```

```python
d = sqrt(b^2 - 4*a*c)
d
```

```python
d.diameter()
```

```python
r1 = (-b + d)/(2*a)
r1.diameter()
```

```python
r2 = (-b - d)/(2*a)
r2.diameter()
```

With 8-bit precision arithmetic the errors on $d$, $r_1$, and $r_2$ are respectively $1 \times 10^{-2}$, $2 \times 10^{-1}$, and $2 \times 10^{-2}$.  The error on the roots cannot be smaller than the error on $d$.  The error on $r_1$ is ten times larger than that of $r_2$.

### answer to part 2 of assignment three

```python
RIF16 = RealIntervalField(prec=16)
a = RIF16(3); b = RIF16(100); c = RIF16(2)
d_16 = sqrt(b^2 - 4*a*c)
print('error on d  :', '%.1e' % d_16.diameter())
r1_16 = (-b + d)/(2*a)
print('error on r1 :', '%.1e' % r1_16.diameter())
r2_16 = (-b - d)/(2*a)
print('error on r2 :', '%.1e' % r2_16.diameter())
```

The upper bound on the error on $d$ is $2 \times 10^{-5}$, which is of course smaller than the error on the roots.  The error on the first root is $2 \times 10^{-2}$, one thousand times larger than the error on the second root, which is $6 \times 10^{-5}$.

```python
RIF32 = RealIntervalField(prec=32)
a = RIF32(3); b = RIF32(100); c = RIF32(2)
d_32 = sqrt(b^2 - 4*a*c)
print('error on d  :', '%.1e' % d_32.diameter())
r1_32 = (-b + d_32)/(2*a)
print('error on r1 :', '%.1e' % r1_32.diameter())
r2_32 = (-b - d_32)/(2*a)
print('error on r2 :', '%.1e' % r2_32.diameter())
```

We observe that the upper bound on the error on the first root is still about one thousand times larger than the error on $d$.  Because the error on the first root is $3 \times 10^{-7}$ and larger than $5 \times 10^{-8}$, we double the precision once more.

```python
RIF64 = RealIntervalField(prec=64)
a = RIF64(3); b = RIF64(100); c = RIF64(2)
d_64 = sqrt(b^2 - 4*a*c)
print('error on d  :', '%.1e' % d_64.diameter())
r1_64 = (-b + d_64)/(2*a)
print('error on r1 :', '%.1e' % r1_64.diameter())
r2_64 = (-b - d_64)/(2*a)
print('error on r2 :', '%.1e' % r2_64.diameter())
```

All errors are clearly less than $5 \times 10^{-8}$.  Thus, 64 bits of precision suffice.

### answer to part 3 of assignment three

```python
show(r1_32.center())
show(r1_64.center())
```

We see that only the first 7 decimal places agree.  Thus, 32 bits do not suffice to have the first root accurate with 8 decimal places.  For the other root, 32 bits suffice already:

```python
show(r2_32.center())
show(r2_64.center())
```
