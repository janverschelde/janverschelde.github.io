# Question 1

Use the first 10 convergents of the continued fraction of
$\cos(\sqrt{11})$
to define the first 10 rational approximations of
$\cos(\sqrt{11})$.

Compute the relative errors of
those 10 rational approximations to verify the accuray.

## answer to question 1

```python
c = cos(sqrt(11))
show(c)
```

```python
f = continued_fraction(c)
f
```

```python
q = [f.convergents()[k] for k in range(10)]
show(q)
```

```python
e = [abs(float(x-c))/abs(float(c)) for x in q]
show(e)
```

```python
['%.2e' % x for x in e]
```

We observe that the relative error drops in each step by at least a factor 2.
