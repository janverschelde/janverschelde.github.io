# Question 1

Compute the polar representation of $2 + 3 I$.

Verify that the polar representation corresponds to $2 + 3 I$.

## answer to question 1

```python
z = 2 + 3*I
show(z)
```

```python
r = abs(z)
r
```

```python
a = arg(z)
a
```

The polar representation of $z$ is given by $\sqrt{13}$ and $\arctan(3/2)$.

```python
y = r*(cos(a) + I*sin(a))
show(y)
```

```python
y - z
```

The expansion of the polar representation leads to $z$.

# Question 2

Does $p = x^2 + 2 x + 3$ factor over ${\mathbb Q}[x]$?

Extend $\mathbb Q$ with a formal root of $p$. 

What is the other root of $p$ in the extended field?

## answer to question 2

```python
reset()
```

```python
p = QQ[x](x^2 + 2*x + 3)
show(p)
```

```python
factor(p)
```

```python
p.is_irreducible()
```

Both the output of ``factor(p)`` and ``p.is_irreducible()`` show that $p$ has no rational roots.

```python
E.<a> = QQ.extension(p)
show(E)
```

```python
factor(E[x](p))
```

After adding $a$ as a root to $\mathbb Q$ and factoring $p$ over the extended field, we see that the other root of $p$ is $-a - 2$.  

Alternatively, we can ask for the roots of $p$ in $E$.

```python
p.roots(E)
```
