# Question 1

For $n$ variables $x_1, x_2, \ldots, x_n$, define the matrix

$$
   S_n = \left[
   \begin{array}{ccccc}
      x_1 & x_2 & x_3 & \cdots & x_n \\
      x_2 & x_1 & x_2 & \cdots & x_{n-1} \\
      x_3 & x_2 & x_1 & \cdots & x_{n-2} \\
      \vdots & \vdots & \vdots & \ddots & \vdots \\
      x_n & x_{n-1} & x_{n-2} & \cdots & x_1
   \end{array}
   \right].
$$

Apply your definition for $n=8$ and compute $S_8$.

## answer to question 1

```python
n = 8
```

```python
X = [var(f"x{k:d}") for k in range(1,n+1)]
X
```

```python
idx = lambda i,j: abs(i-j)
```

```python
s = Matrix(8,8,idx)
show(s)
```

```python
xdx = lambda i,j: X[idx(i,j)]
S = Matrix(8,8,xdx)
show(S)
```

# Question 2

Consider $x(t)$ a function of $t$ as the solution of the
      initial value problem
$$
   \frac{d^2}{d~\!t^2}~\!x(t) - 2 \frac{d}{d~\!t}~\!x(t) - 3 x(t) = 0,
   \quad x(0) = 4, \quad x'(0) = 0.
$$
Compute $x(t)$ using the Laplace transform.

What is the algebraic equation that must be solved?

## answer to question 2

```python
reset()
```

```python
t = var('t')
```

```python
x = function('x')(t)
```

```python
ode = diff(x, t, 2) - 2*diff(x, t) - 3*x
show(ode)
```

```python
s = var('s')
Lode = ode.laplace('t', 's')
show(Lode)
```

```python
T = var('T')
Teq = Lode.subs({laplace(x, t, s): T})
show(Teq)
```

The above equation shows the algebraic equation in the Laplace tranform ``T`` with the initial conditions which has to be solved.

```python
sol = solve(Teq, T)
show(sol)
```

```python
init = {x(t=0): 4, diff(x,t).subs(t=0): 0}
show(init)
```
