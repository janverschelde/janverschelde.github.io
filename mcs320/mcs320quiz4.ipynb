# Question 1

Let 

```python
s = lambda n: float(sum([exp(-(k/n)**2)*sin(2*k/n) for k in range(n)])/n)
```

be a function to approximate 
$\displaystyle \int_{0}^{1} \exp(-x^2) \sin(2x) dx$.

1. Define a Python function ``p`` which does the same as ``s``.

      For correctness, compare the output of ``p(1000)`` and ``s(1000)``.

2. Apply Cython to improve the efficiency of ``p``.

   Time the execution of the Cython function for $n = 10000$,
   compare with timings of ``p``.

## answer to question 1

```python
def p(n):
    return float(sum([exp(-(k/n)**2)*sin(2*k/n) for k in range(n)])/n)
```

```python
s1000 = s(1000)
s1000
```

```python
p1000 = p(1000)
p1000
```

```python
abs(s1000 - p1000)
```

The error is zero.

Let us now Cythonize:

1. Use the ``sin`` and ``exp`` functions of the C ``math.h``.
2. Arguments of ``sin`` and ``exp`` are floats.
3. Declare ``n`` as a long integer.

```python
%%cython
cdef extern from "math.h":
    double sin(double)

cdef extern from "math.h":
    double exp(double)

def c(long n):
    return sum([exp(-(float(k)/n)**2)*sin(2*float(k)/n) for k in range(n)])/n
```

```python
c(1000)
```

```python
s(1000)
```

```python
abs(c(1000) - s(1000))
```

Verifying the correctness, we see that the difference between ``c(1000)`` and ``s(1000)`` is of the same order as the double precision.

```python
timeit('c(10000)')
```

```python
timeit('p(10000)')
```

We observe a huge speedup of the Cython code over the original Python function.

# Question 2

Let $p = 6 x^4 y^3 - x^2 y^6 + x^2 y^3 + x^2 - 2 y^7 - y^3$.

Without retyping $p$, convert $p$ into
$-x^2 y^6 + 6 x^4 y^3 - 2 y^7 + x^2 y^3 - y^3 + x^2$.

## answer to question 2

We start by typing in $p$ as an expression in $x$ and $y$.

```python
x, y = var('x,y')
```

```python
p = 6*x^4*y^3 - x^2*y^6 + x^2*y^3 + x^2 - 2*y^7 - y^3
show(p)
```

Observing $p$ is given in the question in pure lexicographic order, we cast $p$ into the ring of polynomials with integer coefficients, with a lexicographic monomial order.

```python
R.<x, y> = PolynomialRing(ZZ, order='lex')
```

```python
q = R(p)
show(q)
```

Ordering the monomials of $p$ in degree lexicographic order happens by casting $p$ into a polynomial ring with a ``deglex`` monomial order.

```python
S.<x, y> = PolynomialRing(ZZ, order='deglex')
```

```python
r = S(p)
show(r)
```
