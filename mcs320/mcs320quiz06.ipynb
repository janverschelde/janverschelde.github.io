# Question 2

Consider the recursive function 
$$
         f(n, 0) = 1, \quad f(0, m) = 1, \quad
         f(n, m) = 2~\!n~\!f(n-1, m) + 3~\!m~\!f(n, m-1).
$$
      Define a Python function ``F`` which returns $f(n, m)$,
      using a dictionary for memoization.
      
Make sure your ``F`` can compute $f(100, 100)$.

## answer to question 2

```python
reset()
```

```python
def F(n, m, D={}):
    if (n, m) in D:
        return D[(n, m)]
    else:
        if n == 0:
            result = 1
        elif m == 0:
            result = 1
        else:
            result = 2*n*F(n-1, m) + 3*m*F(n, m-1)
        D[(n, m)] = result
        return result
```

```python
F(100, 100)
```
