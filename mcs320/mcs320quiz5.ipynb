# Question 2

Write an efficient recursive function to compute
$$
         p_0(x) = 1, ~~ p_1(x) = x - 1, ~~ \mbox{for } n > 1:
         p_n(x) = (2 x + 1) p_{n-1}(x) + (x - 1) p_{n-2}(x).
$$
Make sure your function can compute $p_{100}(x)$.
Do *NOT* use ``cached_function``.

## answer to question 2

```python
def p(n, x, D={}):
    """
    Memoized computation of p using D.
    """
    if (n, x) in D:
        return D[(n, x)]
    else:
        if n == 0:
            result = 1
        elif n == 1:
            result = x-1
        else:
            result = expand((2*x + 1)*p(n-1,x) + (x-1)*p(n-2,x))
        D[(n, x)] = result
        return result
```

```python
p(0, x)
```

```python
p(1, x)
```

```python
p(2, x)
```

```python
p(100, x)
```
