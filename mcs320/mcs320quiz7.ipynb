# Question 1

Consider $P = (1, 0)$ as a point on the curve
$(x^2 + y^2 + 2y)^2 - (x^2 + y^2) = 0$.

1. Compute the tangent line to the curve at $P$.
2. What is the slope of the tangent line at $P$?

## answer to question 1

```python
x, y = var('x, y')
curve = (x^2 + y^2 + 2*y)^2 - (x^2 + y^2)
show(curve)
```

We compute the Taylor series, truncate after the linear terms.

```python
t = taylor(curve, (x, 1), (y,0), 1)
show(t)
```

The tangent line is the equation $2 x + 4 y - 2$, or in the $y = m x + b$ form:

```python
s = solve(t, y)
show(s)
```

We extract the slope as follows:

```python
s[0].rhs().coefficient(x,1)
```

Or alternatively, extract the slope directly from the truncated Taylor series:

```python
slope = -t.coefficient(x, 1)/t.coefficient(y, 1)
show(slope)
```

# Question 2

Plot the curve defined by $(x^2 + y^2 + 2y)^2 - (x^2 + y^2) = 0$
selecting good ranges for $x$ and $y$.

Convert into polar coordinates and plot.

Describe the difference between the two plots.

## answer to question 2

```python
reset()
```

```python
x, y = var('x, y')
c = (x^2 + y^2 + 2*y)^2 - (x^2 + y^2)
show(c)
```

```python
implicit_plot(c, (x, -2, 2), (y, -3, 1))
```

```python
r, t = var('r, t')
```

```python
cr = c(x=r*cos(t), y=r*sin(t))
show(cr)
```

```python
rt = solve(cr, r)
show(rt)
```

```python
rt0 = rt[0].rhs().trig_simplify()
show(rt0)
```

```python
polar_plot(rt0, (t, 0, 2*pi))
```

In polar coordinates, the curve has a much simpler form and is therefore easier to plot as well.  In particular, we do not need to be concerned about giving ranges for $x$ and $y$.
