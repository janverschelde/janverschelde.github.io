``QQbar`` extends the field ``QQ`` of rational numbers with algebraic numbers.
An algebraic number is defined as a root of polynomial with rational
coefficients.  To avoid the expression swell caused by exact computation
with roots, floating-point interval arithmetic is applied.
Interval arithmetic is the computation with intervals.
The intervals store the lower and upper bounds on the floating-point
number represented by the interval.

# 1. Defining QQbar Numbers

We start defining an element of ``QQbar`` using an example of the documentation.

The field ``QQbar`` extends the field ``QQ`` of rational numbers.  Therefore, we need to start with a polynomial with rational coefficients.  As the example, we take $p = x^5 - x - 1$.  In the code cell below, the expression $x^5 - x - 1$ is cast into the polynomial ring ``QQ[x]``, so that $p$ is a polynomial with rational coefficients.

```python
p = QQ[x](x^5 - x - 1)
type(p)
```

As $p$ has degree 5, it has 5 roots and we have to specify a small enough interval that isolates a root.  Following the documentation, we take $[0.1, 0.2]$ as the interval for the real part and $[1.0, 1.1]$ for the interval of the imaginary part of the root.  The ``CIF`` is an abbreviation for ``ComplexIntervalField`` and ``RIF`` abbreviates ``RealIntervalField``.  In the code cell below the interval ``iv`` is constructed.

```python
iv = CIF(RIF(0.1, 0.2), RIF(1.0, 1.1))
iv
```

The question marks in the short output form of ``iv`` indicate that we have at most once decimal place of accuracy in the interval number.  With the long output form, the ``brackets`` style shows the explicit lower and upper bounds for the complex interval.  

```python
iv.str(style='brackets')
```

And then, after defining the polynomial ``p`` and interval ``iv``, we can define a root ``r``.

```python
r = QQbar.polynomial_root(p, iv)
show(r)
```

```python
r in iv
```

The minimial polynomial of an algebraic number is the polynomial with smallest degree that has that number as a root.  It is available in the method ``minpoly()`` on a element of ``QQbar``.

```python
r.minpoly()
```

The minimal polynomial of ``r`` is the exact representation of the algebraic number ``r``.  The shortest *symbolic verification* evaluates the root ``r`` at ``p``, as done below.

```python
p(x=r)
```

We may view ``r`` as an element of ``CIF``, but only after a typecast.

```python
type(r.real())
```

The command ``r.real().str(style='brackets')`` results in an error because the ``str`` method on an ``AlgebraicReal`` has no ``style`` attribute.  If we cast ``r.real()`` into the ``RealIntervalField`` (abbreviated by ``RIF``), then we can see the interval notation for the real part of the root ``r``.

```python
ivr = RIF(r.real())
type(ivr)
```

```python
ivr.str(style='brackets')
```

The long output form of the algebraic root ``r`` is obtained as follows.

```python
CIF(r).str(style='brackets')
```

The *numerical verification* that ``r`` is a root of ``p`` uses complex interval arithmetic.

```python
p(x=CIF(r))
```

The tiny numbers in the interval on return reflect the default working precision of the ``ComplexIntervalField``.

The fragile step in the construction of an element of ``QQbar`` is the selection of an initial complex interval in the constructor.  Asking for all roots of the polynomial in ``QQbar`` facilitates the construction of this interval.

```python
R = p.roots(QQbar)
R
```

What is key in this definition is the distinction between the symbolic and numeric verification of a constructed QQbar number.

To compute the n-th roots of a number, there is a shortcut.  For example, to compute all 5-th roots of 2, we can do the following:

```python
fifthsqrt2 = QQbar(2).nth_root(5, all=True)
fifthsqrt2
```

```python
fifthsqrt2[0].minpoly()
```

# 2. QQbar Arithmetic

Because we defined ``r`` as a root of the polynomial $x^5 - x - 1$, we have that $r^5 - r - 1 = 0$, or equivalently $r^5$ simplifies into $r+1$.  Let us verify this property.

```python
z1 = r^5
show(z1)
```

```python
z2 = r + 1
show(z2)
```

```python
z1 == z2
```

```python
CIF(z1) == CIF(z2)
```

```python
type(z1)
```

As algebraic numbers, the result of $r^5$ and $r + 1$ are identical to each other.  Their numerical representations as complex intervals are not the same.

Let us compare with the exact way of working with algebraic numbers.

```python
E.<a> = QQ.extension(p)
show(E)
```

The ``E.<a> = QQ.extension(p)`` defined the field ``E`` as the extension of ``Q`` using a root of the polynomial ``p`` denoting this root by ``a``.

```python
a^5
```

```python
r^5
```

The result of the symbolic verification that ``a`` is a root of ``p``, causing ``a^5`` to simplify to ``a+1`` is a lot nicer than the outcome of ``r^5``.

However, the exact symbolic computations may lead to expression swell, as illustrated below.

```python
a^43/(a^32+1/123)
```

```python
r^43/(r^32+1/123)
```

The analogue computation with ``QQbar`` arithmetic results always in two intervals, whereas the coefficients of the fourth degree polynomial ``a`` start growing rapidly.

```python
timeit('a^43/(a^32+1/123)')
```

```python
timeit('r^43/(r^32+1/123)')
```

Remarkably however, the symbolic computations still happen faster than those with ``QQbar``.

## 3. Coordinates of Regular Polygons

The complex roots of unity lie in the complex plane on a circle with radius one and define the corners of a regular polygon.  Consider the pentagon.

```python
pentagon = QQbar(1).nth_root(5, all=True)
pentagon
```

The first complex number has coordinates $(\cos(2 \pi/5), \sin(2 \pi/5))$, verified below:

```python
p1 = [cos(RR(2*pi/5)), sin(RR(2*pi/5))]
p1
```

The exact representation of this first point is defined by its minimal polynomial.

```python
pentagon[1].minpoly()
```

Computing the minimal polynomial for the real and imaginary parts separately ...

```python
(pentagon[1].real().minpoly(), pentagon[1].imag().minpoly())
```

... shows that the real part is the root of a quadratic polynomial.
